```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
from matplotlib import colors
product= ["pro1", "pro2", "pro3", "pro4", "pro5"]
sales = [100, 150, 75, 200, 120]

plt.bar(product, sales)
plt.title("sales data for five record")
plt.xlabel("product")
plt.ylabel("sales")
plt.show()
```

```python
### line chart create monthly temp.
months = ["jan", "feb", "mar", "apr", "may", "jun", "jul", "aug", "sep", "oct", "nov", "dec"]
temp = [10,22,23,25,12,45,65,32,35,37,45,37]

plt.plot(months, temp)
plt.title("monthly temperature changes city")
plt.xlabel("months")
plt.ylabel("temp")
plt.show()
```

```python
### scatter plot to show the relationship between hours studied and exam scores for 20 students.
hours=[1,2,2.5,3,3.5,2.4,2.6,4,4.5,2,3,8,4.5,5,6.5,1.9,2.3,4.5,6,9]
marks=[45,43,47,41,40,45,42,49,42,20,46,48,47,42,46,21,59,53,12,50]

plt.scatter(hours,marks)
plt.title("hours studied and exam scores for 20 students")
plt.xlabel("hours")
plt.ylabel("marks")
plt.show()
```

```python
###pie chart representing the percentage distribution of different fruits sold in a store.
fruits = ["Apple","Banana","Orange","Mango"]
quantile =[20,25,38,40]

plt.pie(quantile,labels=fruits)
plt.title("percentage distribution of different fruits sold in a store")
plt.show()
```

```python
### histogram to display the frequency distribution of ages of 100 people.

ages= np.random.randint(10,80,100)
plt.hist(ages)
plt.title("frequency distribution of ages of 100 people")
plt.xlabel("ages")
plt.ylabel("frequency")
plt.show()
```

```python
###Plot both bar and line chart together showing revenue and profit for 6 months.
months_6 = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
revenue = [500, 600, 550, 700, 750, 800]
profit = [100, 150, 120, 200, 220, 250]

plt.bar(months_6, revenue, color='lightblue', label='Revenue')
plt.plot(months_6, profit, color='red', marker='o', linewidth=2, label='Profit')
plt.title('Revenue and Profit (6 Months)')
plt.xlabel('Months')
plt.ylabel('Amount')
plt.legend()
plt.show()
```

```python
###Create a bar chart with customized colors, labels, and grid lines.
categories = ['A', 'B', 'C', 'D']
values = [40, 70, 55, 90]
colors = ['#FF5733', '#33FF57', '#3357FF', '#F333FF']

plt.bar(categories, values, color=colors)
plt.title('Customized Bar Chart', fontsize=14, fontweight='bold')
plt.xlabel('Categories')
plt.ylabel('Values')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
### line chart and annotate the highest and lowest points.


days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
values = [45, 70, 30, 85, 60, 95, 50]

plt.plot(days, values, marker='o', color='blue')
plt.title('Line Chart with Highest and Lowest Annotations')
plt.xlabel('Days')
plt.ylabel('Values')


max_val = max(values)
max_idx = values.index(max_val)
plt.annotate(f'Highest ({max_val})',
             xy=(days[max_idx], max_val),
             xytext=(days[max_idx], max_val - 15),
             arrowprops=dict(facecolor='green', shrink=0.05))


min_val = min(values)
min_idx = values.index(min_val)
plt.annotate(f'Lowest ({min_val})',
             xy=(days[min_idx], min_val),
             xytext=(days[min_idx], min_val + 12),
             arrowprops=dict(facecolor='red', shrink=0.05))

plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
